```python
from openai import OpenAI
from google.colab import userdata

# Get API key from Colab Secrets
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))

# Store conversation history
# conversation = {
#     "messages": []
# }

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        break

    # Add user's message to conversation
    conversation["messages"].append({
        "role": "user",
        "content": user_input
    })

    # Send entire conversation to OpenAI
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=conversation["messages"],
        temperature=0.7,
        max_tokens=250
    )

    # Extract AI response
    assistant_response = response.choices[0].message.content

    # Add AI response to conversation
    conversation["messages"].append({
        "role": "assistant",
        "content": assistant_response
    })

    print("AI:", assistant_response)
    print("-" * 50)

print("\nConversation History:")
print(conversation)
```

```python
# Cnoversation Summary Memory
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))

# Store only a summary instead of the complete conversation
conversation_summary = ""

while True:

    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        break

    # Create the prompt using only the summary + latest question
    messages = [
        {
            "role": "system",
            "content": """
            You are a helpful conversational AI.
            Use the conversation summary to maintain context.
            Answer the user's latest question clearly and concisely.
            """
        },
        {
            "role": "user",
            "content": f"""
            Previous Conversation Summary:
            {conversation_summary}

            New User Question:
            {user_input}
            """
        }
    ]

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=0.7
    )

    assistant_response = response.choices[0].message.content

    print("\nAI:", assistant_response)

    # Display token usage
    usage = response.usage

    print("\n--- Token Usage ---")
    print("Input Tokens:", usage.prompt_tokens)
    print("Output Tokens:", usage.completion_tokens)
    print("Total Tokens:", usage.total_tokens)

    # Update the conversation summary
    summary_response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": """
                Summarize the conversation in simple, crisp language.
                Keep only information that is useful for maintaining
                future conversational context.
                Do not include unnecessary details.
                """
            },
            {
                "role": "user",
                "content": f"""
                Existing Summary:
                {conversation_summary}

                New User Question:
                {user_input}

                AI Response:
                {assistant_response}
                """
            }
        ],
        temperature=0
    )

    conversation_summary = summary_response.choices[0].message.content

    print("\n--- Current Summary ---")
    print(conversation_summary)
    print("=" * 60)
```

```python
# Store Only the Last N Conversations
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))

conversation = []

MAX_TURNS = 3

while True:

    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        break

    # Add user message
    conversation.append({
        "role": "user",
        "content": user_input
    })

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=conversation,
        temperature=0.7
    )

    assistant_response = response.choices[0].message.content

    # Add AI response
    conversation.append({
        "role": "assistant",
        "content": assistant_response
    })

    print("\nAI:", assistant_response)

    # Token information
    print("\n--- Token Usage ---")
    print("Input Tokens:", response.usage.prompt_tokens)
    print("Output Tokens:", response.usage.completion_tokens)
    print("Total Tokens:", response.usage.total_tokens)

    # Keep only the latest N turns
    conversation = conversation[-(MAX_TURNS * 2):]

    print("\n--- Stored Messages ---")
    print(len(conversation))

    print("=" * 60)
```
